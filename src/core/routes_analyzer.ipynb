```python
import pandas as pd
```

```python
df = pd.read_csv('03-01-2025 BIANCA KELLY DE SOUZA PULTZ.csv', sep=';')
df
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')

str_columns = list(df.dtypes[df.dtypes == 'object'].index)

for col in str_columns:
    df[col] = df[col].str.lower().str.replace(' ', '_')
```

```python
df
```

```python
df.destination_address.value_counts()
```

```python
def combine_packages(series):
    return ', '.join(series)
```

```python
df['packages'] = df.groupby('destination_address')['spx_tn'].transform(combine_packages)
print(df[['destination_address', 'packages']])
```

```python
df_routes = df[['destination_address', 'packages']].drop_duplicates()
print(df_routes)
```

```python
df[['destination_address', 'packages']].drop_duplicates().to_csv('df_routes.xlsx', index=False, encoding='utf-8', sep=';')
```
